# Guía de transición de ipywidgets a tkinter

Esta guía explica cómo transformar la interfaz del minimarket desde `ipywidgets` a `tkinter`. El objetivo es comprender los nuevos componentes, la forma de conectar botones y la organización de una aplicación de escritorio.

La guía puede abrirse en Google Colab, pero las ventanas de `tkinter` deben ejecutarse localmente en Visual Studio Code. Colab sirve aquí para estudiar y copiar el código.

## 1. Diferencia principal

Con `ipywidgets`, la interfaz aparece dentro del notebook y se utiliza `display()` para mostrarla. Con `tkinter`, se crea una ventana independiente que permanece abierta gracias a `mainloop()`.

La lógica del proyecto puede mantenerse: buscar productos, agregar al carrito, calcular el total, registrar una venta y descontar el inventario. Lo que cambia principalmente es la forma de construir la interfaz y conectar los botones.

````python
# En el proyecto usaremos este alias para mantener el nombre widgets
import tkinter as widgets
from tkinter import ttk, messagebox

ventana = widgets.Tk()
ventana.title('Minimarket')
ventana.geometry('700x450')

# La ventana se ejecuta localmente, no en Google Colab
# ventana.mainloop()
````

## 2. Componentes que aparecen en tkinter

`Tk` crea la ventana principal. `Label` muestra textos, `Entry` permite escribir, `Button` crea botones y `Spinbox` permite seleccionar cantidades.

`ttk.Combobox` reemplaza al menú desplegable de `ipywidgets`. `ttk.Treeview` permite mostrar los productos y el carrito en forma de tabla. `LabelFrame` agrupa componentes relacionados y `Text` sirve para mostrar la boleta.

Finalmente, `messagebox` muestra mensajes emergentes, por ejemplo, cuando una venta se registra correctamente.

````python
# Equivalencias principales

# ipywidgets.HTML       -> tkinter.Label
# ipywidgets.Text       -> tkinter.Entry
# ipywidgets.Dropdown   -> ttk.Combobox
# ipywidgets.Button     -> tkinter.Button
# ipywidgets.HBox/VBox  -> tkinter.Frame o LabelFrame
# display()             -> ventana.mainloop()
# on_click()            -> command=funcion
````

## 3. Crear componentes nuevos

En `tkinter`, cada componente recibe primero la ventana o el marco donde será ubicado. Por ejemplo, el botón pertenece a `ventana`. El texto visible se indica con `text`, no con `description`.

````python
import tkinter as widgets
from tkinter import ttk

ventana = widgets.Tk()

etiqueta = widgets.Label(ventana, text='Producto')
entrada = widgets.Entry(ventana)
cantidad = widgets.Spinbox(ventana, from_=1, to=100)
categoria = ttk.Combobox(ventana, values=['Todas', 'Bebidas'])

etiqueta.pack()
entrada.pack()
cantidad.pack()
categoria.pack()

# Para ejecutar localmente:
# ventana.mainloop()
````

## 4. Conectar un botón con una función

En `ipywidgets` se utilizaba `on_click`. En `tkinter`, la función se entrega al argumento `command`. Es importante escribir el nombre de la función sin paréntesis, porque tkinter la ejecutará cuando el usuario presione el botón.

````python
def saludar():
    mensaje.config(text='Botón presionado')

ventana = widgets.Tk()
mensaje = widgets.Label(ventana, text='Presiona el botón')
mensaje.pack(pady=10)

boton = widgets.Button(
    ventana,
    text='Presionar',
    command=saludar
)
boton.pack()

# Ejecutar localmente en VS Code:
# ventana.mainloop()
````

## 5. Mostrar productos en una tabla

`Treeview` se utiliza para mostrar datos en filas y columnas. En el minimarket se usa una tabla para mostrar código, nombre, categoría, precio y stock.

````python
tabla = ttk.Treeview(
    ventana,
    columns=('codigo', 'nombre', 'precio', 'stock'),
    show='headings'
)

tabla.heading('codigo', text='Código')
tabla.heading('nombre', text='Nombre')
tabla.heading('precio', text='Precio')
tabla.heading('stock', text='Stock')

tabla.insert('', 'end', values=('A001', 'Arroz 1 kg', '$1500', 20))
tabla.pack()

# Para saber qué fila seleccionó el usuario:
seleccion = tabla.selection()
# datos = tabla.item(seleccion[0], 'values')
````
